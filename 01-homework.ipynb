```python
median_hpw = data["horsepower"].median()
mode_hpw = data["horsepower"].value_counts().index[0]

data["horsepower"] = data["horsepower"].fillna(mode_hpw)
median_hpw2 = data["horsepower"].median()
print("Before:", median_hpw)
print("After filling NA's:", median_hpw2)
```

```python
X = data.loc[data["origin"]=="Asia", ["vehicle_weight", "model_year"]].iloc[:7, :].values
XTX = np.matmul(np.transpose(X), X)
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
z = np.matmul(np.linalg.inv(XTX), np.transpose(X))
w = np.matmul(z, y)
```
